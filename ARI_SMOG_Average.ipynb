```python
import pandas as pd

# Load the data from the CSV file (assuming you have already loaded the data)
data = pd.read_csv('/content/results_comments_average.csv')  # Replace with your file path

# Calculate the average and standard deviation for ARI and SMOG for both original posts, comments, and responses
averages_std = {
    'ARI Average (STD) for Original Post': (data['ari_original'].mean(), data['ari_original'].std()),
    'SMOG Average (STD) for Original Post': (data['smog_original'].mean(), data['smog_original'].std()),
    'ARI Average (STD) for Comments': (data['avg_ari_comments'].mean(), data['avg_ari_comments'].std()),
    'SMOG Average (STD) for Comments': (data['avg_smog_comments'].mean(), data['avg_smog_comments'].std()),
    'ARI Average (STD) for GPT Response': (data['ari_response'].mean(), data['ari_response'].std()),
    'SMOG Average (STD) for GPT Response': (data['smog_response'].mean(), data['smog_response'].std()),
}

# Output the results
for category, (avg, std) in averages_std.items():
    print(f"{category}: {avg:.2f} ({std:.2f})")
```
