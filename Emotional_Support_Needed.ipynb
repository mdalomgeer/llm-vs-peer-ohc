```python
import openai
import pandas as pd

# Load the dataset
file_path = '/content/Post_Response_title_formated_CodingFormat.xlsx'
df = pd.read_excel(file_path)
```

```python
import os
# Set your OpenAI API key
openai.api_key = os.environ["OPENAI_API_KEY"]
```

```python
def evaluate_emotional_support(response):
    messages = [
        {"role": "system", "content": "You are an AI assistant that evaluates whether emotional support is needed."},
        {"role": "user", "content": f'The response is: "{response}"\n\nIf the response indicates that emotional support is needed, return "1". If not, return "0".'}
    ]

    completion = openai.ChatCompletion.create(
        model="gpt-3.5-turbo",
        messages=messages,
        max_tokens=10,
        temperature=0
    )

    result = completion.choices[0].message['content'].strip()
    return int(result) if result in ["0", "1"] else None
```

```python
!pip install openai==0.28
```

```python
# Apply the function to the 'chatgpt_response' column in the dataset
df['Emotional_Support_Needed'] = df['chatgpt_response'].apply(evaluate_emotional_support)

# Save the results back to an Excel file
output_file = '/content/evaluated_dataset.xlsx'
df.to_excel(output_file, index=False)

print(f"Evaluation completed. Results saved to '{output_file}'.")
```
